```python
#!/usr/bin/env python3
"""
Тестовый пайплайн Ultralytics: сначала короткий прогон (по умолчанию 5 эпох),
затем продолжение с resume до заданного общего числа эпох.

Снимки весов: epoch_0001.pt, epoch_0010.pt, epoch_0020.pt, … (копия last.pt
после соответствующей эпохи). Папка: <project>/<name>/epoch_snapshots/

Зависимости: pip install ultralytics

Пример (Colab / локально):
  python TrainingAlgorithm.py --data dataset/data.yaml --model yolo12n.pt --total-epochs 150
  python TrainingAlgorithm.py --data dataset/data.yaml --total-epochs 80 --warmup-epochs 5

Colab + Google Drive (всё пишется на диск):
  python TrainingAlgorithm.py --mount-google-drive --data /content/.../data.yaml --total-epochs 150
  # при необходимости своя папка на диске:
  python TrainingAlgorithm.py --mount-google-drive --drive-base /content/drive/MyDrive/MyRuns ...
"""
from __future__ import annotations

import argparse
import shutil
import sys
from pathlib import Path


DEFAULT_LOCAL_PROJECT = Path("runs/detect")
DEFAULT_DRIVE_BASE = Path("/content/drive/MyDrive/SirenixYOLO")


def mount_google_drive(mount_point: str = "/content/drive") -> None:
    """Смонтировать Google Drive в Colab (иначе ImportError)."""
    try:
        from google.colab import drive  # type: ignore
    except ImportError as e:
        print(
            "ERROR: пакет google.colab недоступен (нужна среда Google Colab).\n"
            "  Локально уберите --mount-google-drive или установите пути вручную.",
            file=sys.stderr,
        )
        raise SystemExit(1) from e
    print(f"[drive] mounting Google Drive at {mount_point} …", flush=True)
    drive.mount(mount_point, force_remount=False)
    print("[drive] ready.", flush=True)


def _parse_args() -> argparse.Namespace:
    p = argparse.ArgumentParser(
        description="YOLO: warmup-эпохи, затем resume до total-epochs; снимки каждые 10 эпох (+ эпоха 1)."
    )
    p.add_argument(
        "--data",
        required=True,
        type=Path,
        help="Путь к data.yaml (train/val, names).",
    )
    p.add_argument(
        "--model",
        default="yolo12n.pt",
        help="Базовые веса для первой фазы (например yolo12n.pt, yolo11n.pt).",
    )
    p.add_argument(
        "--total-epochs",
        type=int,
        default=150,
        help="Итоговое число эпох (включая warmup). Вторая фаза дотренирует до этого числа.",
    )
    p.add_argument(
        "--warmup-epochs",
        type=int,
        default=5,
        help="Первая фаза без resume (проверка пайплайна).",
    )
    p.add_argument(
        "--mount-google-drive",
        action="store_true",
        help="Смонтировать Google Drive (Colab) и по умолчанию писать runs туда.",
    )
    p.add_argument(
        "--drive-base",
        type=Path,
        default=DEFAULT_DRIVE_BASE,
        help="Базовая папка на диске: в неё попадёт runs/detect/… (только с --mount-google-drive, если не задан --project).",
    )
    p.add_argument(
        "--project",
        type=Path,
        default=None,
        help="Корень project Ultralytics. Если не указан: локально runs/detect; с --mount-google-drive — <drive-base>/runs/detect.",
    )
    p.add_argument(
        "--name",
        default="popup_resume_test",
        help="Имя прогона (подпапка в project).",
    )
    p.add_argument(
        "--imgsz",
        type=int,
        default=640,
        help="Размер входа.",
    )
    p.add_argument(
        "--batch",
        type=int,
        default=16,
        help="Batch size (-1 для auto).",
    )
    p.add_argument(
        "--device",
        default="",
        help="Пусто = auto; для Colab GPU обычно 0.",
    )
    p.add_argument(
        "--workers",
        type=int,
        default=8,
        help="DataLoader workers.",
    )
    p.add_argument(
        "--seed",
        type=int,
        default=0,
        help="Seed для воспроизводимости.",
    )
    ns = p.parse_args()
    if ns.project is not None:
        ns.project = ns.project.expanduser().resolve()
    else:
        if ns.mount_google_drive:
            ns.project = (ns.drive_base.expanduser().resolve() / "runs" / "detect")
        else:
            ns.project = DEFAULT_LOCAL_PROJECT.resolve()
    return ns


def _snapshot_dir(project: Path, name: str) -> Path:
    return project / name / "epoch_snapshots"


def _make_epoch_snapshot_callback(snapshot_root: Path):
    """
    После сохранения чекпоинта копируем last.pt в epoch_XXXX.pt для эпох 1, 10, 20, …
    Используется on_fit_epoch_end (после validate и save_model), чтобы last.pt уже был актуален.
    Номер эпохи — 1-based, как в логах Ultralytics.
    """

    def on_fit_epoch_end(trainer) -> None:
        ep = int(trainer.epoch) + 1
        if ep != 1 and ep % 10 != 0:
            return
        wdir = Path(trainer.wdir)
        last = wdir / "last.pt"
        if not last.is_file():
            return
        snapshot_root.mkdir(parents=True, exist_ok=True)
        dst = snapshot_root / f"epoch_{ep:04d}.pt"
        shutil.copy2(last, dst)
        print(f"[snapshot] epoch {ep} -> {dst}", flush=True)

    return on_fit_epoch_end


def _train_kwargs_common(args: argparse.Namespace, resume: bool) -> dict:
    kw: dict = {
        "data": str(args.data.resolve()),
        "project": str(args.project.resolve()),
        "name": args.name,
        "imgsz": args.imgsz,
        "batch": args.batch,
        "workers": args.workers,
        "seed": args.seed,
        "resume": resume,
    }
    if args.device:
        kw["device"] = args.device
    return kw


def main() -> int:
    args = _parse_args()

    if args.mount_google_drive:
        mount_google_drive("/content/drive")
        args.project.mkdir(parents=True, exist_ok=True)
        print(f"[drive] Ultralytics project (runs root): {args.project}", flush=True)

    if not args.data.is_file():
        print(f"ERROR: data.yaml не найден: {args.data}", file=sys.stderr)
        return 1

    if args.total_epochs <= 0:
        print("ERROR: --total-epochs должен быть > 0", file=sys.stderr)
        return 1

    if args.warmup_epochs <= 0:
        print("ERROR: --warmup-epochs должен быть > 0", file=sys.stderr)
        return 1

    if args.total_epochs <= args.warmup_epochs:
        print(
            f"ERROR: --total-epochs ({args.total_epochs}) должен быть больше "
            f"--warmup-epochs ({args.warmup_epochs}), иначе вторая фаза не нужна.",
            file=sys.stderr,
        )
        return 1

    try:
        from ultralytics import YOLO
    except ImportError as e:
        print(f"ERROR: ultralytics: {e}\n  pip install ultralytics", file=sys.stderr)
        return 1

    snap = _snapshot_dir(args.project, args.name)
    callback = _make_epoch_snapshot_callback(snap)

    # --- фаза 1: с нуля, без resume ---
    print(f"[phase 1] train {args.warmup_epochs} epochs from {args.model} -> resume test")
    model = YOLO(args.model)
    model.add_callback("on_fit_epoch_end", callback)

    kw1 = _train_kwargs_common(args, resume=False)
    kw1["epochs"] = args.warmup_epochs
    model.train(**kw1)

    weights_dir = Path(args.project) / args.name / "weights"
    last_pt = weights_dir / "last.pt"
    if not last_pt.is_file():
        print(f"ERROR: после фазы 1 нет {last_pt}", file=sys.stderr)
        return 1

    # --- фаза 2: продолжение до total_epochs ---
    print(f"[phase 2] resume from {last_pt} up to total epoch {args.total_epochs}")
    model2 = YOLO(str(last_pt))
    model2.add_callback("on_fit_epoch_end", callback)

    kw2 = _train_kwargs_common(args, resume=True)
    kw2["epochs"] = args.total_epochs
    model2.train(**kw2)

    best = weights_dir / "best.pt"
    print("Done.")
    print(f"  weights: {weights_dir}")
    print(f"  snapshots: {snap}")
    if best.is_file():
        print(f"  best: {best}")
    return 0


if __name__ == "__main__":
    raise SystemExit(main())
```
